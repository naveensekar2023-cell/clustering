clustering

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("/energy_data.csv")
df.head()
df
```

data cleaninig and preprocessing

```python
df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
df = df.set_index('timestamp')

df.isnull().sum()
```

select feature and normalize the data

```python
X = df[['energy_consumption', 'temperature']]
scaler = StandardScaler()
X_scaled=scaler.fit_transform(X)
X_scaled
```

```python
smart_city = []
for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, init='k-means++', random_state=42)
    kmeans.fit(X_scaled)
    smart_city.append(kmeans.inertia_)

# Plotting the Elbow Method
plt.figure(figsize=(10, 6))
plt.plot(range(1, 11), smart_city, marker='*', linestyle='solid')
plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('Smart Cities') # Note: Usually this represents 'Inertia' or 'WCSS'
plt.show()
```

```python
best_k = 3
kmeans = KMeans(n_clusters=best_k, init='k-means++', random_state=42)
df['cluster'] = kmeans.fit_predict(X_scaled)
df
```

evaluate the model performance

```python
from sklearn.metrics import silhouette_score
df['cluster'] = kmeans.labels_
silhouette_avg = silhouette_score(X_scaled,df['cluster'])
print(f'silhouette score:{silhouette_avg:.2f}')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the figure size (10 inches wide, 6 inches tall)
plt.figure(figsize=(10, 6))

# Create the scatter plot
sns.scatterplot(
    x='energy_consumption',
    y='temperature',
    hue='cluster',
    data=df,
    palette='viridis',
    s=30
)

# Add titles and labels
plt.title('Clusters of Energy Consumption Patterns')
plt.xlabel('Energy Consumption')
plt.ylabel('Temperature')
plt.legend(title='Cluster')

# Display the plot
plt.show()
```
